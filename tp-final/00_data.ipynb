```python
# Hay 13 columnas con valores faltantes. Algunos campos pueden tenerlos, es natural que una reseña venga sin un comentario adjunto.
#Las fechas correspondientes a la tabla de orders "olist_orders_dataset" puede estar relacionado a algun aspecto comercial.. Quizas no se entrego todavia
# En "olist_products_dataset" se ve un patron, 610 registros que no tienen información y 2 que no tienen unidades de medida fisica  

from pyspark.sql import functions as F

resultados_nulos = []

for nombre, df in datasets.items():
    total = df.count()

    conteos = df.agg(*[
        F.sum(
            F.when(
                F.col(c).isNull() |
                (F.trim(F.col(c).cast("string")) == ""),
                1
            ).otherwise(0)
        ).alias(c)
        for c in df.columns
    ]).first().asDict()

    for columna, cantidad in conteos.items():
        cantidad = int(cantidad or 0)

        if cantidad > 0:
            resultados_nulos.append((
                nombre,
                columna,
                total,
                cantidad,
                round(100 * cantidad / total, 2)
            ))

df_nulos = spark.createDataFrame(
    resultados_nulos,
    [
        "tabla",
        "columna",
        "total_filas",
        "valores_faltantes",
        "porcentaje_faltante"
    ]
)

display(
    df_nulos.orderBy(F.desc("porcentaje_faltante"))
)
```

```python
#Integridad del resto de relaciones. Todas acordes.
from pyspark.sql import functions as F

# Relaciones a verificar:
# (tabla_origen, columna_origen, tabla_destino, columna_destino)

relaciones = [
    (
        "olist_order_items_dataset", "product_id",
        "olist_products_dataset", "product_id"
    ),
    (
        "olist_orders_dataset", "customer_id",
        "olist_customers_dataset", "customer_id"
    ),
    (
        "olist_order_items_dataset", "order_id",
        "olist_orders_dataset", "order_id"
    ),
    (
        "olist_order_items_dataset", "seller_id",
        "olist_sellers_dataset", "seller_id"
    ),
    (
        "olist_order_payments_dataset", "order_id",
        "olist_orders_dataset", "order_id"
    ),
    (
        "olist_order_reviews_dataset", "order_id",
        "olist_orders_dataset", "order_id"
    )
]

resultados = []

for tabla_origen, clave_origen, tabla_destino, clave_destino in relaciones:

    origen = datasets[tabla_origen]
    destino = datasets[tabla_destino]

    # Detectar referencias inexistentes
    huerfanos = origen.join(
        destino.select(
            F.col(clave_destino).alias(clave_origen)
        ).distinct(),
        on=clave_origen,
        how="left_anti"
    )

    total_origen = origen.count()
    total_huerfanos = huerfanos.count()

    # Distinguir claves nulas de referencias inexistentes
    claves_nulas = origen.filter(
        F.col(clave_origen).isNull()
    ).count()

    referencias_inexistentes = total_huerfanos - claves_nulas

    resultados.append((
        tabla_origen,
        clave_origen,
        tabla_destino,
        total_origen,
        claves_nulas,
        referencias_inexistentes,
        round(100 * total_huerfanos / total_origen, 4)
        if total_origen > 0 else 0.0
    ))

df_integridad = spark.createDataFrame(
    resultados,
    [
        "tabla_origen",
        "clave",
        "tabla_destino",
        "total_registros",
        "claves_nulas",
        "referencias_inexistentes",
        "porcentaje_sin_correspondencia"
    ]
)

display(df_integridad)
```

```python
#Consistencia temporal. El análisis identificó 1.359 pedidos (1,3666 %) cuya fecha de despacho es anterior a la fecha de aprobación registrada, así como 23 pedidos (0,0231 %) cuya fecha de entrega es anterior a la fecha de despacho.


from pyspark.sql import functions as F

orders = datasets["olist_orders_dataset"]

# Convertimos las columnas a timestamp para compararlas
fechas = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]

orders_fechas = orders

for columna in fechas:
    orders_fechas = orders_fechas.withColumn(
        columna,
        F.to_timestamp(F.col(columna))
    )

# Definimos las reglas de consistencia temporal
reglas = {
    "aprobacion_antes_compra": (
        F.col("order_approved_at") <
        F.col("order_purchase_timestamp")
    ),
    "despacho_antes_aprobacion": (
        F.col("order_delivered_carrier_date") <
        F.col("order_approved_at")
    ),
    "entrega_antes_despacho": (
        F.col("order_delivered_customer_date") <
        F.col("order_delivered_carrier_date")
    ),
    "entrega_antes_compra": (
        F.col("order_delivered_customer_date") <
        F.col("order_purchase_timestamp")
    )
}

# Contamos las anomalías por regla
resultados = []

for nombre_regla, condicion in reglas.items():
    cantidad = orders_fechas.filter(condicion).count()

    resultados.append((
        nombre_regla,
        cantidad,
        round(cantidad / orders_fechas.count() * 100, 4)
    ))

df_temporal = spark.createDataFrame(
    resultados,
    [
        "regla",
        "pedidos_inconsistentes",
        "porcentaje_sobre_total"
    ]
)

display(df_temporal)
```

```python

# Identificar pedidos con al menos una anomalía temporal

orders_anomalias = orders_fechas.withColumn(
    "flag_inconsistencia_temporal",
    (
        F.coalesce(reglas["aprobacion_antes_compra"], F.lit(False)) |
        F.coalesce(reglas["despacho_antes_aprobacion"], F.lit(False)) |
        F.coalesce(reglas["entrega_antes_despacho"], F.lit(False)) |
        F.coalesce(reglas["entrega_antes_compra"], F.lit(False))
    )
)

resultado = orders_anomalias.agg(
    F.count("*").alias("total_pedidos"),
    F.sum(
        F.col("flag_inconsistencia_temporal").cast("int")
    ).alias("pedidos_con_anomalias")
)

display(resultado)
```

```python

from pyspark.sql import functions as F

geo = datasets["olist_geolocation_dataset"]

# Cantidad de registros por prefijo postal
geo_por_prefijo = (
    geo.groupBy("geolocation_zip_code_prefix")
    .agg(F.count("*").alias("cantidad_registros"))
)

# Resumen de granularidad y duplicación
total_registros = geo.count()
prefijos_distintos = geo.select(
    "geolocation_zip_code_prefix"
).distinct().count()

duplicados_exactos = total_registros - geo.distinct().count()

prefijos_multiples = geo_por_prefijo.filter(
    F.col("cantidad_registros") > 1
).count()

max_registros_prefijo = geo_por_prefijo.agg(
    F.max("cantidad_registros")
).first()[0]

resultado = spark.createDataFrame(
    [(
        total_registros,
        prefijos_distintos,
        duplicados_exactos,
        prefijos_multiples,
        max_registros_prefijo
    )],
    [
        "total_registros",
        "prefijos_distintos",
        "duplicados_exactos",
        "prefijos_con_multiples_registros",
        "max_registros_por_prefijo"
    ]
)

display(resultado)
```

```python
#Se investigaron los ocho prefijos postales asociados a más de un estado en el dataset de geolocalización.

#En todos los casos se identificó un estado ampliamente predominante y una única observación correspondiente a un estado alternativo.

#Este comportamiento sugiere posibles inconsistencias en la asignación geográfica, aunque no permite confirmar por sí solo que los registros minoritarios sean incorrectos.

#Para la capa Silver se propone conservar las observaciones originales, aplicar deduplicación exacta y construir una representación consolidada por prefijo postal.
from pyspark.sql import functions as F

geo = datasets["olist_geolocation_dataset"]

variabilidad_geo = (
    geo.groupBy("geolocation_zip_code_prefix")
    .agg(
        F.countDistinct("geolocation_city").alias("ciudades_distintas"),
        F.countDistinct("geolocation_state").alias("estados_distintos"),
        F.countDistinct(
            F.struct("geolocation_lat", "geolocation_lng")
        ).alias("coordenadas_distintas")
    )
)

resultado = variabilidad_geo.agg(
    F.count("*").alias("total_prefijos"),
    F.sum(
        F.when(F.col("ciudades_distintas") > 1, 1).otherwise(0)
    ).alias("prefijos_multiples_ciudades"),
    F.sum(
        F.when(F.col("estados_distintos") > 1, 1).otherwise(0)
    ).alias("prefijos_multiples_estados"),
    F.sum(
        F.when(F.col("coordenadas_distintas") > 1, 1).otherwise(0)
    ).alias("prefijos_multiples_coordenadas")
)

display(resultado)


from pyspark.sql import functions as F

geo = datasets["olist_geolocation_dataset"]

# Identificar prefijos asociados a más de un estado
prefijos_conflictivos = (
    geo.groupBy("geolocation_zip_code_prefix")
    .agg(
        F.countDistinct("geolocation_state").alias("estados_distintos")
    )
    .filter(F.col("estados_distintos") > 1)
    .select("geolocation_zip_code_prefix")
)

# Mostrar distribución de estados para esos prefijos
detalle_conflictos = (
    geo.join(
        prefijos_conflictivos,
        on="geolocation_zip_code_prefix",
        how="inner"
    )
    .groupBy(
        "geolocation_zip_code_prefix",
        "geolocation_state"
    )
    .agg(
        F.count("*").alias("cantidad_registros"),
        F.countDistinct("geolocation_city").alias("ciudades_distintas")
    )
    .orderBy(
        "geolocation_zip_code_prefix",
        F.desc("cantidad_registros")
    )
)

display(detalle_conflictos)
```

```python
# Análisis de validez de importes monetarios y cuotas
#Se evalúa la calidad de los campos monetarios de los datasets olist_order_items_dataset y olist_order_payments_dataset, considerando los precios de productos, costos de envío y valores de pago.

#Adicionalmente, se analiza la cantidad de cuotas registrada en los pagos.

#El objetivo es identificar valores nulos, negativos, iguales a cero y posibles valores extremos que requieran una revisión posterior.

#Las validaciones se realizan sin modificar los datos originales. Los resultados permitirán definir reglas de calidad para la capa Silver y establecer criterios confiables para la construcción de indicadores financieros en Gold.



from pyspark.sql import functions as F

items = datasets["olist_order_items_dataset"]
payments = datasets["olist_order_payments_dataset"]

# Campos que vamos a analizar
campos = [
    ("order_items", items, "price"),
    ("order_items", items, "freight_value"),
    ("order_payments", payments, "payment_value"),
    ("order_payments", payments, "payment_installments")
]

resultados = []

for tabla, df, columna in campos:

    # Conversión para evaluar los valores numéricos
    valor = F.col(columna).cast("double")

    resumen = df.agg(
        F.count("*").alias("total"),
        F.sum(F.when(F.col(columna).isNull(), 1).otherwise(0)).alias("nulos"),
        F.sum(
            F.when(
                F.col(columna).isNotNull() & valor.isNull(),
                1
            ).otherwise(0)
        ).alias("no_numericos"),
        F.sum(F.when(valor < 0, 1).otherwise(0)).alias("negativos"),
        F.sum(F.when(valor == 0, 1).otherwise(0)).alias("ceros"),
        F.min(valor).alias("minimo"),
        F.max(valor).alias("maximo")
    ).first()

    resultados.append((
        tabla,
        columna,
        resumen["total"],
        resumen["nulos"],
        resumen["no_numericos"],
        resumen["negativos"],
        resumen["ceros"],
        resumen["minimo"],
        resumen["maximo"]
    ))

df_validez_monetaria = spark.createDataFrame(
    resultados,
    [
        "tabla",
        "campo",
        "total_registros",
        "nulos",
        "no_numericos",
        "negativos",
        "ceros",
        "minimo",
        "maximo"
    ]
)

display(df_validez_monetaria)
#Los resultados permiten descartar que la ausencia de entrega sea la explicación principal de los costos de envío iguales a cero.

#Estos valores podrían corresponder a envíos gratuitos, promociones u otras condiciones comerciales, aunque no es posible determinar su causa exacta a partir de la información disponible.

#En consecuencia, se propone conservar los registros con freight_value = 0 en Silver, considerándolos valores potencialmente válidos. Se implementarán controles para detectar costos negativos o no numéricos, sin reemplazar arbitrariamente los valores iguales a cero.


from pyspark.sql import functions as F

items = datasets["olist_order_items_dataset"]
orders = datasets["olist_orders_dataset"]

resultado = (
    items
    .filter(F.col("freight_value") == 0)
    .join(
        orders.select("order_id", "order_status"),
        on="order_id",
        how="left"
    )
    .groupBy("order_status")
    .agg(
        F.count("*").alias("lineas_envio_gratis"),
        F.countDistinct("order_id").alias("pedidos_distintos")
    )
    .orderBy(F.desc("lineas_envio_gratis"))
)

display(resultado)

#Análisis de pagos con valores atípicos
#En Olist, un mismo pedido (order_id) puede tener varios registros de pago en order_payments.
#Por ejemplo, un cliente puede pagar una compra combinando tarjeta de crédito y vouchers.

from pyspark.sql import functions as F

payments = datasets["olist_order_payments_dataset"]
orders = datasets["olist_orders_dataset"]

# Pagos con importe cero o cantidad de cuotas cero
pagos_especiales = (
    payments
    .filter(
        (F.col("payment_value") == 0) |
        (F.col("payment_installments") == 0)
    )
    .join(
        orders.select("order_id", "order_status"),
        on="order_id",
        how="left"
    )
    .select(
        "order_id",
        "payment_sequential",
        "payment_type",
        "payment_installments",
        "payment_value",
        "order_status"
    )
    .orderBy("payment_value", "payment_installments")
)

display(pagos_especiales)



```
